```python
# Join inventory_master table to products table to form an inventory_health table. 

### Code ###
inventory_health = data['inventory_master'].merge(data['products'], on ='product_id', suffixes = ('_inv', '_prod'))
inventory_health.head(10)
```

```python
# join the inventory_health table to the branches table to form a full master table.
### code ###
inventory_health = inventory_health.merge(data['branches'], on = 'branch_id', suffixes = ('', '_branch'))
inventory_health.head()
```

```python
inventory_health['CAUTION_current_stock_unreliable'] = (inventory_health['current_stock'] > inventory_health['max_stock'])
inventory_health['CAUTION_current_stock_unreliable'].value_counts()
```

```python
### Data Quality Flag: current_stock Anomaly

##Issue:
#'current_stock' values are implausibly large relative to 'max_stock' across the entire dataset - not isolated rows.

##Evidence:
#'current_stock' range: 88853 - 121013 (avg ~107240)
#'max_stock' range: 131 - 582 (avg ~322)
#100% of rows (180/180) have 'current_stock > max_stock'

##Likely cause:
# Units mismatch or data-generation error - no warehouse realistically holds 100,000+ units against a stated capacity of ~500.

##Action taken:
# Flagged via 'CAUTION_current_stock_unreliable' column rather than silently corrected. Does **not** affect other columns (opening_stock, reorder_level, safety_stock, max_stock all fall within plausible ranges).

##Decision needed from team:
# Should 'current_stock' be corrected with an assumed scale factor, excluded from analysis, or is there a source-data explanation the team is aware of? Do not build stockout/overstock KPIs on this column until resolved.
```

```python
inventory_health.duplicated().sum()
```

```python
import re

str_cols = inventory_health.select_dtypes(include='object').columns

inventory_health[str_cols] = inventory_health[str_cols].apply(
    lambda x: x.str.replace(r'[\x00-\x1f\x7f]', '', regex=True)   # CLEAN: strip non-printable chars
                .str.replace(r'\s+', ' ', regex=True)               # collapse multiple spaces to one
                .str.strip()                                        # TRIM: remove leading/trailing space
)

inventory_health.sample(10)
```

```python
inventory_health.to_csv('inventory_health_master.csv', index=False)
```
